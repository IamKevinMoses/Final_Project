```python
import pandas as pd
import numpy as np
```

```python
#Load the dataset
df = pd.read_csv("C:/Users/ASUS/Desktop/ICBT Assignments/1. New Batch Assignments/Final Project/Final_Project/data/raw/water_pollution_disease.csv",keep_default_na=False
)
```

```python
#Create a copy
clean_df = df.copy()
```

```python
#Standardize column names
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
#Rename long columns to short names
clean_df.rename(columns={
    "contaminant_level_(ppm)": "contaminant_level",
    "ph_level": "ph",
    "turbidity_(ntu)": "turbidity",
    "dissolved_oxygen_(mg/l)": "dissolved_oxygen",
    "nitrate_level_(mg/l)": "nitrate",
    "lead_concentration_(µg/l)": "lead",
    "bacteria_count_(cfu/ml)": "bacteria_count",
    "access_to_clean_water_(%_of_population)": "clean_water_access",
    "diarrheal_cases_per_100,000_people": "diarrheal_cases",
    "cholera_cases_per_100,000_people": "cholera_cases",
    "typhoid_cases_per_100,000_people": "typhoid_cases",
    "infant_mortality_rate_(per_1,000_live_births)": "infant_mortality_rate",
    "gdp_per_capita_(usd)": "gdp_per_capita",
    "healthcare_access_index_(0-100)": "healthcare_access",
    "urbanization_rate_(%)": "urbanization",
    "sanitation_coverage_(%_of_population)": "sanitation_coverage",
    "rainfall_(mm_per_year)": "rainfall",
    "temperature_(°c)": "temperature",
    "population_density_(people_per_km²)": "population_density"
}, inplace=True)
```
