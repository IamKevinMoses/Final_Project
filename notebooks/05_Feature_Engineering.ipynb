```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
#Load the final dataset
df = pd.read_csv(
    "../data/final/waterborne_disease_risk_dataset.csv",
    keep_default_na=False
)
df.head()
```

```python
#Select the Target Variable
y = df["risk_level"]

print(y.head())
```

```python
#Remove the target leakage
X = df.drop(columns=[
    "risk_level",
    "diarrheal_cases",
    "cholera_cases",
    "typhoid_cases",
    "total_disease_cases"
])
```

```python
print(X.columns.tolist())
```

```python
#Encode categorical variables
categorical_columns = X.select_dtypes(include="object").columns

print(categorical_columns)
```

```python
X = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)
```

```python
print("Feature Matrix Shape:", X.shape)
```

```python
#Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

```python
#Feature scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
#Verify scaling
print(type(X_train_scaled))
print(X_train_scaled.shape)

print(type(X_test_scaled))
print(X_test_scaled.shape)
```

```python
#Save everything
X_train.to_csv("../data/model_input/X_train.csv", index=False)

X_test.to_csv("../data/model_input/X_test.csv", index=False)

y_train.to_csv("../data/model_input/y_train.csv", index=False)

y_test.to_csv("../data/model_input/y_test.csv", index=False)
```

```python
#Save the scaler
import joblib

joblib.dump(
    scaler,
    "../models/scaler.pkl"
)
```
